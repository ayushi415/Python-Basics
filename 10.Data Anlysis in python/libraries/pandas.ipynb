```python
### Series
# A pandas series is a one_dimentional array-like object that can hold any data type. It is similar to a column in a table.

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("Series :\n",series )
print(type(series))
```

```python
## Create a Series from dictionary
import pandas as pd
data={'a': 1,'b': 2,'c': 3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
### Dataframe
## A Dataframe is a two-dimensional,size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

# Create a Dataframe from a dictionary of list

data={
    'Name' : ['Govind','Vasudev','Madhav'],
    'Age' : [21,20,19],
    'City' : ['Mumbai','Ahemdabad','Bangalore']  
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Create a Dataframe from a List of Dictionaries

data=[
    {'Name':'Govind', 'Age':32 , 'City':'Vrindavan'},
     {'Name':'Sweety', 'Age':20 , 'City':'Ahemdabad'},
      {'Name':'Radha', 'Age':19 , 'City':'Mathura'},
       {'Name':'Ayushi', 'Age':22 , 'City':'Goa'}
]
df=pd.DataFrame(data)
print(df)
print(type(df))
    
    
```
